```python
# Student Result Analyzer


# Step 1: Get student name
while True:
    name = input("Enter student name: ").strip()

    if name:
        break
    else:
        print("Student name cannot be empty.")


# Step 2: Define subjects
subjects = [
    "Python",
    "Java",
    "Database",
    "Web Development",
    "Mathematics"
]


# Step 3: Get marks
marks = {}

for subject in subjects:

    while True:
        try:
            mark = float(input(f"Enter marks for {subject} (0-100): "))

            # Step 4: Validate marks
            if 0 <= mark <= 100:
                marks[subject] = mark
                break
            else:
                print("Marks must be between 0 and 100.")

        except ValueError:
            print("Please enter a valid number.")


# Step 5: Calculate total
total = sum(marks.values())


# Step 6: Calculate percentage
percentage = total / len(subjects)


# Step 7: Determine grade
if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"


# Step 8: Determine PASS/FAIL
passed_all_subjects = all(mark >= 35 for mark in marks.values())

if passed_all_subjects:
    result = "PASS"
else:
    result = "FAIL"


# Step 9: Display result
print("\n======================================")
print("         STUDENT RESULT")
print("======================================")

print(f"Student Name: {name}")

print("--------------------------------------")

for subject, mark in marks.items():
    print(f"{subject:<20}: {mark:.2f}")

print("--------------------------------------")

print(f"Total Marks : {total:.2f} / {len(subjects) * 100}")
print(f"Percentage  : {percentage:.2f}%")
print(f"Grade       : {grade}")
print(f"Result      : {result}")

print("======================================")
```
